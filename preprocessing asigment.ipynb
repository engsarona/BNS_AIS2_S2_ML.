```python
import os
import pandas as pd


def Read_data_file(file_path):
    
    try:
        if not os.path.exists(file_path):
            print(f"Error: The file path '{file_path}' does not exist.")
            return None

        df = pd.read_csv(file_path)
        print("Data loaded successfully.")
        return df

    except Exception as e:
        print(f"Error reading the file: {e}")
        return None


def Drop_unnecessary_features(df, cols_to_drop):
   
    if df is None:
        print("DataFrame is empty or not loaded.")
        return None

    existing_cols = [col for col in cols_to_drop if col in df.columns]
    df_cleaned = df.drop(columns=existing_cols)
    print(f"Dropped columns: {existing_cols}")
    return df_cleaned


def Check_data_type(df):

    if df is None:
        print("DataFrame is empty or not loaded.")
        return None

    report = {
        "Column Name": [],
        "Data Type": [],
        "Unique Values Count": []
    }

    for col in df.columns:
        report["Column Name"].append(col)
        report["Data Type"].append(df[col].dtype)
        report["Unique Values Count"].append(df[col].nunique())

    report_df = pd.DataFrame(report)
    
    report_df = report_df.set_index("Column Name").T
    return report_df
```
