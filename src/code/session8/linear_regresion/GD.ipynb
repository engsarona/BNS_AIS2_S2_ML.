```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Data
x = np.array([1,2,3,4])
y = np.array([2,2,8,3,6,4,5])
n = len(x)
# initailization parameters
w = 0
b = 0
#print(w,b)
#0.63*x + 0.9 = y'
```

```python
#Hyperparamters
learning_rate = 0.01
epochs = 100
cost_values = []
print(f"inital W : {w}",
initial b :{b},
learning_rate : {learning_rate},
epoches : {epochs})
print("traning ")
```

```python
learning_rate = 0.01
epochs = 100
cost_values = []
print(f"inital W : {w}\n"
 f"inital b : {b}\n"
 f"learning rate: {learning_rate}\n"
 f"Epoches: {epochs}"
 )
print("training....................")
```

```python
#Gradient Desent
#===============================
for i in range (epochs) :
    #Prediction
    y_hat = w * x + b
    #error
    error = y_hat - y
    #Gradients
    D_w = np.mean(error* x)
    D_b = np.mean(error)
    #update parameters
    w = w - learning_rate* D_w
    b = b - learning_rate* D_b
    y_hat = w*x + b 
    #cost function
    cost = 0.5 * np.mean((y_hat - y)**2)


    if (i + 1) % 100 ==0 or i ==0 :
        print(f"Epoch {i + 1}: ")
        print(f"Cost =  {cost}: ")
```

```python


x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 4.0, 6.0, 8.0])

w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000

for i in range(epochs):
    y_hat = w * x + b
    error = y_hat - y
    
    D_w = np.mean(error * x)
    D_b = np.mean(error)
    
    w = w - learning_rate * D_w
    b = b - learning_rate * D_b
    
    cost = 0.5 * np.mean(error ** 2)
    
    if (i + 1) % 100 == 0 or i == 0:
        print(f"Epoch {i + 1}: Cost = {cost:.6f} | w = {w:.4f} | b = {b:.4f}")
```

```python
plt.figure(figsize=(8,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x, w * x + b , 'r' )
plt.title("Linear Regression")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.subplot(1,2,2)
plt.plot(cost_values )
plt.title("cost vs Epochs")
plt.xlabel("Epoches")
plt.ylabel("Cost")
```
