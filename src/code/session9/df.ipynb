```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_csv('Salary_Data.csv' )
```

```python
data.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.pairplot(data, x_vars=['YearsExperience'], y_vars=['Salary'], height=3)
plt.show()
```

```python
# تقسيم البيانات إلى متطلبات (x) وهدف (y) ثم تقسيمها لتصفيات تدريب واختبار
x = data.iloc[:, :-1]
y = data.iloc[:, 1]
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=20)
```

```python
# إنشاء النموذج وتدريبه وحساب التوقعات لبيانات التدريب
lin_reg = LinearRegression()
lin_reg.fit(x_train, y_train)
y_hat_train = lin_reg.predict(x_train)
```

```python
# عرض القيم المتوقعة (y_hat_train)
y_hat_train
```

```python
# y = w * x + b

plt.scatter(x_train, y_train, color='red')
plt.plot(x_train, y_hat_train, color='green')
plt.title("linear Regression ")
plt.xlabel("years")
plt.ylabel("Salary")
plt.show()
```

```python
y_hat_test = lin_reg.predict(x_test)
```

```python
print("linear regression eval")
print("=" * 40)
print("MSE: ", mean_squared_error(y_train, y_hat_train))
```
