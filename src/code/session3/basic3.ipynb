```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"Training Model: {model_name}")

    print("\n Datasets")
    for dataset in datasets:
        print("-",dataset)

    print("\nHyperparameters")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")
```

```python
train_model(
    "Fraud Detection",
    "train.csv",
    "validation.csv",
    "test.csv",
    learning_rate=0.001,
    epochs=100,
    batch_size=64,
    optimizer="Adan",
    early_stopping=True
)
```
