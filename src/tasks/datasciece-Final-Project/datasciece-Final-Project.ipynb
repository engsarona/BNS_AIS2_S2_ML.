## <mark>The above result show some missing values exist in some columns So let's display them </mark>

```python
#yOUR CODES
missing_values = df_clean.isnull().sum()
missing_percentage = (df_clean.isnull().sum() / len(df_clean)) * 100


missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Percentage (%)': missing_percentage.round(2)
})

missing_info[missing_info['Missing Count'] > 0]
```



> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with ........... for numeric and ......for categorical

```python
# trying to fill missing values in my datset using for loop 

for col in df_clean.columns:
    if df_clean[col].isnull().sum() > 0:
        if df_clean[col].dtype in ['int64', 'float64']:
            
            median_val = df_clean[col].median()
            df_clean[col] = df_clean[col].fillna(median_val)
        elif df_clean[col].dtype.name in ['object', 'category']:
            
            mode_val = df_clean[col].mode()[0]
            df_clean[col] = df_clean[col].fillna(mode_val)


print("Remaining Missing Values:")
df_clean.isnull().sum()[df_clean.isnull().sum() > 0]

```

```python
#check the Nulls 
df_clean.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
# fill  dataset with missing values.
    

df_clean['member_gender'] = df_clean['member_gender'].fillna('Unknown')


print("Remaining Missing Values:")
df_clean.isnull().sum()
```

```python
##  step  2 ) Incorrect data type as in (( `user_type` &&	`member_gender`)) :  significant change , so  need to clean it.
gbf['user_type'] = gbf['user_type'].astype('category')
gbf['member_gender'] = gbf['member_gender'].astype('category')


print(gbf[['user_type', 'member_gender']].dtypes)
```

```python
## check your work 

gbf['member_birth_year'] = gbf['member_birth_year'].astype(int)
gbf['member_age'] = 2019 - gbf['member_birth_year']


gbf[['member_birth_year', 'member_age']].describe()
```

```python
## step 3 ) Good factor for marketing work as in ((`member_birth_year`)) : so need to clean it.
## To know the target audience's age and serve on them

# create a column for users' age depending on the current year


median_birth_year = gbf['member_birth_year'].median()
gbf['member_birth_year'] = gbf['member_birth_year'].fillna(median_birth_year).astype(int)


gbf['user_age'] = 2019 - gbf['member_birth_year']


print(gbf[['member_birth_year', 'user_age']].describe())
```

```python
## check your work 
print(gbf[['member_birth_year', 'user_age']].info())
print("\nAge Summary Statistics:")
print(gbf['user_age'].describe())
```

```python
## change the new column age data type from float to int

gbf['user_age'] = gbf['user_age'].astype(int)


print(gbf['user_age'].dtype)
print(gbf[['member_birth_year', 'user_age']].head())
```

```python
gbf[['user_age']].info()
```

```python
## create a function to know the average users ages
def get_average_user_age(df):
    
    avg_age = df['user_age'].mean()
    print(f"Average User Age: {avg_age:.2f} years")
    return avg_age


get_average_user_age(gbf)
```

```python
## another important factor to offer  marketing services is the trip duration time 

## need to change the time from second to minute to be more usable

gbf['duration_min'] = (gbf['duration_sec'] / 60).round(2)


print(gbf[['duration_sec', 'duration_min']].head())
print("\nTrip Duration (Minutes) Summary:")
print(gbf['duration_min'].describe())

```

```python
## I need to change the time from second to Houre  to be more usable

gbf['duration_hour'] = (gbf['duration_sec'] / 3600).round(2)


print(gbf[['duration_sec', 'duration_min', 'duration_hour']].head())
print("\nTrip Duration (Hours) Summary:")
print(gbf['duration_hour'].describe())
```
