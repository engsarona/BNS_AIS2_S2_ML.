```python
scores = {
    "student1": {"math": 80, "science": 90},
    "student2": {"math": 75, "science": 85},
    "student3": {"math": 92, "science": 88},
}

total = 0      

for student_data in scores.values():
    for score in student_data.values(): 
        total += score              

print(total)
```
