```python
import numpy as np
import matplotlib.pyplot as plt

class LinearRegressionGD:
    def __init__(self, learning_rate=0.0001, n_iters=1000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0.0
        self.theta_1 = 0.0
        self.sse_history = []

    def fit(self, X, y):
        n = len(X)
        for _ in range(self.n_iters):
            y_pred = self.theta_0 + self.theta_1 * X
            
            
            sse = np.sum((y - y_pred) ** 2)
            self.sse_history.append(sse)
            
            
            d_t0 = (-2 / n) * np.sum(y - y_pred)
            d_t1 = (-2 / n) * np.sum((y - y_pred) * X)
            
            # Update Weights
            self.theta_0 -= self.lr * d_t0
            self.theta_1 -= self.lr * d_t1

    def predict(self, X):
        return self.theta_0 + self.theta_1 * X

    def plot_training(self, X, y):
        plt.figure(figsize=(10, 4))
        
        
        plt.subplot(1, 2, 1)
        plt.plot(self.sse_history, color='blue')
        plt.title('SSE over Iterations')
        plt.xlabel('Iteration')
        plt.ylabel('SSE')

        
        plt.subplot(1, 2, 2)
        plt.scatter(X, y, color='red', label='Data Points')
        plt.plot(X, self.predict(X), color='green', label='Regression Line')
        plt.title('Data & Fitted Line')
        plt.legend()
        plt.show()
```

```python

X = np.array([50, 60, 70, 80, 90])
y = np.array([150, 180, 210, 240, 270])


model = LinearRegressionGD(learning_rate=0.0001, n_iters=1000)
model.fit(X, y)


print(f"Theta 0: {model.theta_0:.4f}")
print(f"Theta 1: {model.theta_1:.4f}")
```

```python

pred_70 = model.predict(70)
print(f"Predicted price for 70 m^2: {pred_70:.2f}")
```

```python

model.plot_training(X, y)
```
